# Resokit.datasets Tutorial


This tutorial is a guide to use ResoKit.datasets package.
ResoKit.datasets provides tools for loading the available datasets:
- [NASA](https://exoplanetarchive.ipac.caltech.edu/)
- [EU](https://exoplanet.eu/home/)

In this tutorial we show, step by step, how to get started with the ResoKit.datasets package.

---

```python
import resokit
import resokit.datasets as rdata

import matplotlib.pyplot as plt

# Load the dataset
data = rdata.load_dataset("eu", check_age=True)
data
```

```python
# If necessary, you can also save the dataset into memory, to avoid reading the source file again
rdata.load_dataset("eu", store=True)
```

```python
eu = rdata.load_dataset("eu")
eu

nasa_full = rdata.load_dataset("nasa", store=True, check_age=True)

# Filter the dataset
nasa = nasa_full[(nasa_full.pl_controv_flag == 0) & (nasa_full.default_flag == 1)]
nasa
```

```python
plt.figure(figsize=(5,4), dpi=130)
plt.plot(nasa.pl_massj, nasa.pl_radj, ".")
plt.semilogx()
plt.semilogy()
plt.xlabel("Planet mass (Jupiter mass)")
plt.ylabel("Planet radius (Jupiter radius)")
plt.suptitle("NASA Exoplanet Archive")
plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, dpi=130, figsize=(10, 4), sharey=True)
axs[0].plot(nasa.pl_orbsmax, nasa.pl_orbeccen, ".", ms=2)
axs[1].plot(eu.semi_major_axis, eu.eccentricity, ".", ms=2)

axs[0].set_title("NASA Exoplanet Archive")
axs[1].set_title("Exoplanet.eu")

for ax in axs.flat:
    ax.set_xscale("log")
    ax.set_xlabel("Semi-major axis (AU)")
axs[0].set_ylabel("Eccentricity")
plt.semilogy()

fig.tight_layout()
plt.show()
```

```python
data = resokit.utils.mass_radius.estimate_mass(
    radius=nasa["pl_radj"],
    radius_err_min=nasa["pl_radjerr1"],
    radius_err_max=nasa["pl_radjerr2"],
    multivariate=[0.1, 0.85],
    )
data
```
