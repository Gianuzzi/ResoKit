# Resokit.datasets Tutorial


This tutorial is a guide to use ResoKit.datasets package.
ResoKit.datasets provides tools for loading the available datasets:
- [NASA](https://exoplanetarchive.ipac.caltech.edu/)
- [EU](https://exoplanet.eu/home/)

In this tutorial we show, step by step, how to get started with the ResoKit.datasets package.

---

## Import ResoKit and the necessary packages

```python
import resokit
import resokit.datasets as datasets

import matplotlib.pyplot as plt
```

## Load the dataset

Let's start with EU catalogue.

```python
data = datasets.load_dataset("eu")
data
```

If necessary, you can also save the dataset into memory, to avoid reading the source file again

```python
datasets.load_dataset("eu", store=True)
```

Now, the dataset will be loaded from memory, instead of reading the file again.

```python
eu = datasets.load_dataset("eu")
eu
```

```python
# Load NASA dataset too
nasa_full = datasets.load_dataset("nasa", store=True, check_age=True)

# Nasa dataset contains all exoplanet solutions (including non default and controversial), so we filter them here
nasa = nasa_full[(nasa_full.controversial == 0) & (nasa_full.default_set == 1)]
nasa
```

We can see both datasets have different amount of exoplanets:

```python
print(f"# of planets in NASA dataset: {nasa.shape[0]}")
print(f"# of planets in EU dataset: {eu.shape[0]}")
```

Checking the columns, we find that both data frames have almost all equal column names




```python
print(f"{'Column':^19s} || {'NASA':^4s} | {'EU':^4s}")
print(f"{"-"*50}")
for col in nasa.columns.sort_values():
    print(f"{col:>19s} || {'YES':^4s} | {'YES' if col in eu.columns else 'NO':^4s}")
```

The only differences are:
- controversial : The exoplanet is confirmed (0) or not yet (1).
- default_set : The solution is the default set (1) or not (0).
- n_planets : Amount of planets in that system.
- n_stars : Amount of stars in that system.

We can see both datasets are similar, but not equal.

```python
plt.figure(figsize=(5,4), dpi=130)
plt.plot(eu.mass, eu.radius, ".", ms=10, label="EU")
plt.plot(nasa.mass, nasa.radius, "x", ms=3, label="NASA")
plt.semilogx()
plt.semilogy()
plt.xlabel("Planet mass [M$_\\mathrm{J}$]")
plt.ylabel("Planet radius [R$_\\mathrm{J}$]")
plt.suptitle("Mass vs Radius comparison")
plt.xlim(1e-4, 1e2)
plt.ylim(bottom=1e-2)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5,4), dpi=130)
plt.plot(eu.a, eu.e, ".", ms=10, label="EU")
plt.plot(nasa.a, nasa.e, "x", ms=3, label="NASA")
plt.semilogx()
plt.semilogy()
plt.xlabel("Planet semimajor-axis [AU]")
plt.ylabel("Planet eccentricity")
plt.suptitle("$a$ vs $e$ comparison")
plt.ylim(bottom=1e-5)
plt.legend()
plt.tight_layout()
plt.show()
```

## Load full dataset

Some planet and star data are not shown because the full data has been trimmed, when using `to_resokit=True` in `load_dataset` function.

All data can be obtained using `to_resokit=True`. Nevertheless, in this case the column names will be the original from each database, instead of being mapped into more comprehensive ones.

```python
nasa_original = datasets.load_dataset("nasa", to_resokit=False)
nasa_original
```

We see that te amount of columns is larger than before

```python
print("# of columns in nasa (as_resokit=True):", nasa.shape[1])
print("# of columns in nasa (as_resokit=False):", nasa_original.shape[1])
```

Also, we can see that **mass** or **radius** are not in the dataframe, but it has **`pl_massj`** and **`pl_radj`** instead.

```python
print("mass" in nasa_original)
print("radius" in nasa_original)

print("pl_massj" in nasa_original)
print("pl_radj" in nasa_original)
```

## Default units

More information about the units can be found at:
- EU: [https://exoplanet.eu/readme/](https://exoplanet.eu/readme/)
- NASA: [https://exoplanetarchive.ipac.caltech.edu/docs/API_PS_columns.html](https://exoplanetarchive.ipac.caltech.edu/docs/API_PS_columns.html)

but the basics are:

- Planet:
    - mass: Jupiter masses
    - radius: Jupiter radii
    - semimajor-axes: au

- Star:
    - mass: Solar masses
    - radius: Solar radii
    - distance: Parsecs
